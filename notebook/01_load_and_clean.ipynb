## rows count before cutoff

```python
before_cutoff_stk = stocks.count()
print("-------Before cutoff---------")
print("")
print(before_cutoff_stk)
print("-------------------------------------------")
print("")
print("====Before Cutoff indexes====")
before_cutoff_idx = indexes.count()
before_cutoff_idx
```

```python
# Date Cutoff
stocks_f = stocks[stocks["Date"].between("2008-01-01","2021-12-31")].copy()
print("Stocks Date cutoff from 2008-01-01 to 2021-12-31")
after_cutoff_stk = stocks_f.count()
print("-----------Stocks After Cutoff---------------")
print("", after_cutoff_stk)
indexes_f = indexes[indexes["Date"].between("2008-01-01","2021-12-31")].copy()
print("Indexes Date cutoff from 2008-01-01 to 2021-12-31")
after_cutoff_idx = indexes_f.count()
print("-----------Indexes After Cutoff---------------")
print("", after_cutoff_idx)

```
